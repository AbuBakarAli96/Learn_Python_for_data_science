## 1. Why do we talk about "environments"?

Python is an **interpreter**: a program that reads your `.py` files and runs
them. But the interpreter is only the beginning. Almost every real project
also uses **third-party packages** — collections of code other people wrote
that you can download and reuse (for example `pandas` for data tables or
`matplotlib` for charts).

All of these packages must be installed *somewhere*, and this is where
problems start:

* Project A needs version 2.0 of a library, project B needs version 3.1.
* Installing everything into one shared place is called the **global**
  Python installation. If you break it, every project on your computer
  can break.
* You want your project to work the same way on your laptop, on your
  friend's laptop, and on the university lab machines.

The solution is a **virtual environment**: a private, copy of Python plus its
own package folder, used by exactly one project.

## 5. pip — installing packages

**pip** is Python's standard package installer. It downloads packages from
the **Python Package Index (PyPI)** — a huge online catalog at
<https://pypi.org> — and installs them into the *current* Python
environment.

We write `python -m pip ...` instead of plain `pip ...`. Both work, but
`python -m pip` guarantees the pip that belongs to **exactly the Python you
just named** — a very common source of confusion is having several Pythons
installed and installing a package into the wrong one.

### The essential pip commands

```bash
# Which pip, and for which Python?
python -m pip --version

# Install a package (latest version):
python -m pip install package_name

# Install a specific version:
python -m pip install package_name==1.2.3

# Upgrade a package:
python -m pip install --upgrade package_name

# List every package installed in this environment:
python -m pip list

# Same list in requirements.txt format:
python -m pip freeze

# Remove a package:
python -m pip uninstall package_name

# Install everything a project needs, from its requirements file:
python -m pip install -r requirements.txt
```

`uninstall` asks "Proceed (y/n)?" — answer `y`, or add `-y` to skip the
question: `python -m pip uninstall -y package_name`.

## 6. Virtual environments with `venv`

`venv` is built into Python — no installation needed. One command creates a
folder (by convention named `.venv`) containing a private Python.

### Create the environment (one time per project)

Windows:

```bash
python -m venv .venv
```

macOS/Linux:

```bash
python3 -m venv .venv
```

### Activate it (every time you open a new terminal)

**Windows (PowerShell):**

```powershell
.venv\Scripts\Activate.ps1
```

**Windows (Command Prompt):**

```bat
.venv\Scripts\activate.bat
```

**macOS / Linux (bash or zsh):**

```bash
source .venv/bin/activate
```

### How do I know it worked?

Your prompt gains a prefix: `(.venv) C:\Users\you\project>` or
`(.venv) you@laptop project %`. Confirm with:

```bash
python --version
python -m pip --version
```

The pip path should now point *inside* `.venv`, e.g.
`...\.venv\lib\site-packages\pip ...`.

PowerShell may block scripts by default. If activation fails with
"running scripts is disabled", run once:

```powershell
Set-ExecutionPolicy -ExecutionPolicy RemoteSigned -Scope CurrentUser
```

### Deactivate (leaving the environment)

```bash
deactivate
```

The `(.venv)` prefix disappears. Deleting the `.venv` folder deletes the
environment completely — it can always be recreated from `requirements.txt`.

### Using the environment with Jupyter

Jupyter must run *inside* the environment to use it. Two common options:

**Option A — simple:** activate the environment, then install and start
Jupyter from it:

```bash
python -m pip install jupyterlab
jupyter lab
```

**Option B — register the environment as a kernel:** with the environment
active, install `ipykernel` and register it; then pick "Python 3 (my-project)"
in Jupyter's kernel menu:

```bash
python -m pip install ipykernel
python -m ipykernel install --user --name my-project --display-name "Python 3 (my-project)"
```

## 7. requirements.txt — recording what a project needs

A `requirements.txt` is a plain text file listing the packages a project
needs, one per line. Anyone (including future-you) can recreate the exact
environment with one command. That is why we say it makes a project
**reproducible**.

Example content:

```text
jupyterlab>=4.0
ipykernel>=6.0
```

Install everything in it:

```bash
python -m pip install -r requirements.txt
```

Save what you currently have installed:

```bash
python -m pip freeze > requirements.txt
```

Version specifiers you will meet:

| Specifier | Meaning |
|---|---|
| `pandas` | any version (not recommended for reproducibility) |
| `pandas==2.2.2` | exactly this version |
| `pandas>=2.0` | this version or newer |
| `pandas>=2.0,<3.0` | a safe range |

For this course the file can be very **short**: Modules 1–6 use only
Python's standard library, so it only needs the Jupyter packages required to
*run this notebook*.

## 8. uv — a modern, faster alternative

[uv](https://docs.astral.sh/uv/) (made by Astral, the makers of the Ruff
code linter) is a recent tool that does the job of pip **and** venv, written
in Rust and much faster. Beginners can happily stay with pip; uv is worth
knowing because many tutorials and workplaces now use it.

Install uv (official installer):

```bash
# Windows (PowerShell):
powershell -ExecutionPolicy ByPass -c "irm https://astral.sh/uv/install.ps1 | iex"

# macOS/Linux:
curl -LsSf https://astral.sh/uv/install.sh | sh
```

Check it:

```bash
uv --version
```

### Workflow 1 — uv as a pip replacement (same venv idea)

`uv pip` is a drop-in interface that mimics pip's commands:

```bash
uv venv                     # create .venv (uses a system Python, or downloads one)
uv pip install package_name # install into .venv (uv finds the environment itself)
uv pip install -r requirements.txt
uv pip list
uv pip uninstall package_name
```

You still activate the environment the normal way before using plain
`python` inside it:

```bash
# macOS/Linux:
source .venv/bin/activate
# Windows PowerShell:
.venv\Scripts\Activate.ps1
```

### Workflow 2 — uv project mode (its own way of managing a project)

In project mode uv keeps dependencies in a `pyproject.toml` file, records
exact versions in a `uv.lock` lockfile, and manages the environment for you:

```bash
uv init my-project      # create a new project folder with pyproject.toml
cd my-project
uv add package_name     # add a dependency (updates pyproject.toml + uv.lock)
uv remove package_name  # remove it again
uv run python script.py # run a command inside the project environment
```

With `uv run` you do not need to activate anything — uv ensures the command
runs inside the project's `.venv` with the locked dependencies.

### pip vs uv — which one for a beginner?

| Situation | Use |
|---|---|
| Following this course, first semester | pip + venv (simplest, always available) |
| Want much faster installs | uv pip workflow |
| Starting a new project where you like lockfiles | uv project mode |
| A tutorial tells you to run `pip install ...` | pip works; `uv pip install ...` also works |

Both tools use the same package format, so switching between them is easy.

## 1. What is programming?

A **computer** is a very fast but very literal machine. It can add millions
of numbers in a blink, but it cannot "figure out" what you *mean* — you must
describe every step exactly.

**Programming** is the craft of writing those step-by-step descriptions in a
language a computer can execute. A person who writes them is a
**programmer** (or developer). The activity itself is called
**programming** or **coding**.

Think of programming as writing a recipe:

* A recipe lists ingredients and steps; a program lists **data** and
  **instructions**.
* A recipe must be followed in order; so is a program.
* A recipe that says "add some sugar" fails; a program must be told
  *exactly* how much, of what, and when.

The difference: a cook can improvise. A computer never does — it does
**exactly** what you wrote, not what you *wanted*.

## 2. What is a program? What is an algorithm?

### Program

A **program** is a sequence of instructions, written in a programming
language, that a computer executes to perform a specific task.

### Algorithm

An **algorithm** is the idea *behind* a program: a finite, precise sequence
of steps that solves a problem — written in plain language first, code later.

**Example algorithm — making tea:**

1. Boil water.
2. Put a tea bag in a cup.
3. Pour the boiling water into the cup.
4. Wait 3 minutes.
5. Remove the tea bag.

Notice the properties of a good algorithm:

* **Finite** — it ends after a limited number of steps.
* **Precise** — every step is unambiguous ("wait 3 minutes", not "wait a bit").
* **Ordered** — steps happen in sequence; you cannot pour water you have
  not boiled.
* **Effective** — each step can actually be carried out.

### A first mathematical algorithm: finding the average of three numbers

1. Get the three numbers, call them `a`, `b`, `c`.
2. Add them: `total = a + b + c`.
3. Divide the total by 3: `average = total / 3`.
4. Report `average`.

The same algorithm becomes a *program* in any language. Here it is in
Python — run the cell (Ctrl+Enter or the ▶ button):

````python
# A tiny program that averages three numbers.
number_a = 7        # first number
number_b = 4        # second number
number_c = 10       # third number

total = number_a + number_b + number_c   # step 2: add them
average = total / 3                      # step 3: divide by three

print("Total:", total)         # step 4a: report the total
print("Average:", average)     # step 4b: report the average
````

### Explanation

Line by line:

* `number_a = 7` — store the value 7 under the name `number_a` (a *variable*;
  fully explained in Module 2).
* `total = number_a + number_b + number_c` — compute the sum and store it.
* `average = total / 3` — compute the average and store it.
* `print(...)` — display results on the screen.
* Lines starting with `#` are **comments**: notes for humans, ignored by
  Python.

Change `number_a` to `10` in the cell above, run it again, and watch the
output change. This edit–run–observe cycle is the heart of programming.

## 3. Problem solving using computers

Real programming is mostly **problem solving**. A reliable approach:

1. **Understand** the problem. What goes in? What should come out?
2. **Plan** the algorithm in plain words or a diagram.
3. **Translate** the plan into code, a small piece at a time.
4. **Test** with known examples.
5. **Fix** mistakes (called *bugs*) and test again. This is *debugging*.

**Worked example** — Problem: "Convert a length in kilometres to miles."

1. *Understand:* input = kilometres; output = miles; 1 km ≈ 0.621371 miles.
2. *Plan:* read kilometres → multiply by 0.621371 → show result.
3. *Translate:* see the next cell.
4. *Test:* 10 km should give ≈ 6.21371 miles.
5. *Fix:* compare against a known value and correct if needed.

````python
# Problem: convert kilometres to miles.
# Plan:    miles = kilometres * 0.621371

kilometres = 10                    # input: the distance in kilometres
conversion_factor = 0.621371       # 1 kilometre in miles

miles = kilometres * conversion_factor   # the conversion step

print(kilometres, "kilometres is", miles, "miles")
````

## 4. Programming languages — high-level vs low-level

Computers only truly understand **machine code**: long strings of numbers
(in binary) that are different for each processor family. Nobody wants to
write programs that way, so we use programming languages at different
"levels":

| | Low-level languages | High-level languages |
|---|---|---|
| Examples | Machine code, Assembly | Python, Java, C++, JavaScript |
| Closeness to hardware | Very close | Far from hardware |
| Readability | Very hard for humans | Designed for humans |
| Speed of execution | Very fast | Usually a bit slower |
| Portability | Tied to one processor type | Runs on many systems |
| Lines for the same task | Many | Few |

**Assembly example** (x86, printing nothing useful to us here, just to
illustrate) — the same "add two numbers" idea:

```asm
mov eax, 5
add eax, 3
```

**Python equivalent:**

```python
result = 5 + 3
```

High-level languages are **translated** into machine code in one of two ways:

* **Compiled** (C, C++, Rust): a *compiler* translates the whole program to
  a machine-code file once; you then run that file. Fast execution, one
  extra build step.
* **Interpreted** (Python, JavaScript): an *interpreter* reads and runs your
  source code directly, statement by statement. Instant feedback — great for
  learning. Python actually compiles your code to an intermediate form
  (called *bytecode*) behind the scenes, but you never manage that yourself.

Python is a **high-level, general-purpose, interpreted** language.

## 6. The Python interpreter, source code, and execution

* **Source code** — the text you write: instructions in a programming
  language, stored in a file such as `hello.py`. It is just text.
* **Interpreter** — the program that reads source code and *executes* it.
  When you "run Python", you are really running the interpreter with your
  file as its input.

### How Python executes a program

```text
hello.py  ──▶  Python interpreter  ──▶  bytecode  ──▶  Python Virtual Machine (PVM)
(source code)   (reads & checks)        (compact        (executes the bytecode
                                         instructions)   instruction by instruction)
```

1. The interpreter **parses** your source code, checking the syntax.
2. It **compiles** it to bytecode (a compact internal instruction format).
3. The **Python Virtual Machine (PVM)** executes the bytecode, statement by
   statement, top to bottom.

If step 1 finds a **syntax error** (e.g. a missing bracket), the program
stops immediately with an error message and *nothing* runs. If the syntax is
fine but an instruction cannot be carried out (e.g. dividing by zero),
execution stops at that point with a **runtime error** — you will meet both
kinds soon.

## 1. Variables

### Concept

A **variable** is a *named box* in the computer's memory where a program
stores a value. Once a value has a name, you can use that name everywhere
instead of repeating the value — and change the value in one place.

In Python, creating a variable is a single **assignment**:

### Syntax

```python
variable_name = value
```

* The `=` sign is the **assignment operator**. It does *not* mean "equals"
  as in maths. It means: compute the value on the **right**, then store it
  under the name on the **left**.

### Example 1 — Beginner

````python
# Create three variables.
student_name = "Amina"     # a text value
age = 19                   # a whole number
quiz_score = 8.5           # a decimal number

# Use them:
print(student_name)
print(age)
print(quiz_score)
````

### Explanation

* `student_name = "Amina"` — the right side `"Amina"` is evaluated, then
  stored under the name `student_name`.
* The order matters: **right side first, then store left.** Writing
  `5 = age` is an error, because `5` is not a name.

### Example 2 — Practical: reassigning variables

A variable can receive a new value at any time; the old value is replaced:

````python
# A bank balance starts at 5000 and later changes.
balance = 5000
print("Starting balance:", balance)

balance = 4250            # reassignment: the 5000 is gone
print("Balance after withdrawal:", balance)

balance = balance - 750   # use the CURRENT value to compute the new one
print("Balance after second withdrawal:", balance)
````

The line `balance = balance - 750` reads naturally if you remember "right
first": Python takes the current `balance` (4250), subtracts 750 → 3500, and
stores that back into `balance`.

### Example 3 — Challenge: swapping two variables

A classic. We want to exchange the values of `first` and `second`:

````python
first = "red"
second = "blue"
print("Before:", first, second)

# Step 1: remember the value of `first` in a spare variable.
temporary = first
# Step 2: copy `second` into `first`.
first = second
# Step 3: restore the saved value into `second`.
second = temporary

print("After: ", first, second)
````

### Naming rules and conventions

**Rules** (Python enforces — breaking them is an error):

1. Only letters, digits and `_`.
2. Cannot start with a digit.
3. Cannot be a keyword (`if`, `for`, `class`, ...).
4. Case-sensitive: `total` and `Total` are different.

```python
valid_names   = 1    # fine
_hidden       = 2    # fine (leading underscore is allowed)
player2       = 3    # fine (digits not at the start)
# 2player      = 4  # SyntaxError: starts with a digit
# my-name      = 5  # SyntaxError: hyphen not allowed
# total marks  = 6  # SyntaxError: space not allowed
# class        = 7  # SyntaxError: keyword
```

**Conventions** (PEP 8 style — not enforced, but expected):

* `snake_case` for variables: `total_price`, `first_name`
* `UPPER_CASE` for constants: `PI = 3.14159`
* Descriptive names: `marks_obtained`, not `m` or `x2`
* Never reuse Python's built-in names (`print`, `str`, `type`, ...) — it
  makes the real tool unusable for the rest of the program.

### Common Mistakes

### Multiple assignment

Python can assign to several names in one line:

````python
# Give three variables their values in one statement:
width, height, depth = 3, 4, 5

# Give several variables the SAME value:
x = y = z = 0

print(width, height, depth)   # 3 4 5
print(x, y, z)                # 0 0 0
````

### Constants by convention

Python has no true constants — but programmers write names in

````python
# Constants: fixed values of the program.
DAYS_IN_WEEK = 7
MAX_MARKS = 100
PI = 3.14159

print("A week has", DAYS_IN_WEEK, "days.")
print("Maximum marks:", MAX_MARKS)
````

Python will not stop you from assigning to `DAYS_IN_WEEK = 8` — the capital

### Under the hood: names and objects (`id()`)

A variable is really a **name that points to an object** in memory. The built-in
`id()` function shows the identity number of the object a name points to. Equal
small numbers share one object; changing a value creates a *new* object and
re-points the name:

````python
a = 10
b = 10
print(id(a), id(b))   # the SAME number twice: both names point to one object

b = b + 1             # 11 is a different object, so b is re-pointed to it
print(id(a), id(b))   # a is unchanged; b now has a different id
````

The exact numbers differ on every computer and every run — only the *pattern*
matters (same, same → then different). This is also why `is` can seem to work for
small numbers (Python reuses them) and why you should still compare values
with `==`.

## 2. Dynamic typing

In languages like C or Java a variable's type is fixed when you declare it.
**Python is dynamically typed**: the *value* carries the type, not the
name. The same name may hold an integer now and text later — Python simply
asks each value what it is.

````python
result = 42          # `result` currently refers to an int
print(type(result))

result = "forty-two" # the same name now refers to a str
print(type(result))
````

Freedom comes with responsibility: dynamic typing means a misspelled or
mixed-up variable is only discovered **when the program runs**. Choose clear
names and keep each variable's purpose consistent.

## 3. Data types

Every value in Python has a **data type**. The fundamental types:

|---|---|---|---|

### Example — `float` and its surprises

* For **displaying**, round to a sensible number of digits (Module 3 shows

* Never compare floats with `==` when a small tolerance is acceptable.

Booleans appear in every decision a program makes. They also behave like

## 4. Type conversion

Programs constantly turn one type into another: text from `input()` must
become a number (Module 3), a number must become text to print inside a
sentence, and so on. The built-in functions do this:

| Function | Converts to | Examples that work | Examples that fail |
|---|---|---|---|
| `int()` | integer | `int("42")`, `int(3.9)`, `int(True)` | `int("3.9")`, `int("abc")` |
| `float()` | float | `float("3.9")`, `float(7)`, `float(True)` | `float("abc")` |
| `str()` | string | `str(42)`, `str(3.9)`, `str(True)` | (almost never fails) |
| `bool()` | boolean | `bool(0)`, `bool("")`, `bool(42)` | (almost never fails) |
| `complex()` | complex | `complex(2)`, `complex("2+3j")` | `complex("2 + 3j")` (extra spaces) |

````python
print(int(3.9))          # float -> int: decimals are CUT, not rounded
print(int("42"))         # str -> int: works when the text is a whole number
print(float(7))          # int -> float
print(float("3.9"))      # str -> float
print(str(42))           # int -> str
print(str(3.9))          # float -> str
print(bool(0), bool(42)) # numbers -> bool: 0 is False, anything else True
print(bool(""), bool("hi"))
print(complex(2))        # int -> complex
````

| `//` | floor division (round down) | `7 // 2` | `3` |
| `%` | modulo (remainder) | `7 % 2` | `1` |
| `**` | exponent (power) | `7 ** 2` | `49` |

Two special ones need explanation:

* `/` **always** produces a float — even `8 / 2` gives `4.0`.
* `//` floors toward negative infinity: `7 // 2` → `3`, but `-7 // 2` → `-4`.
* `%` gives the remainder after division: 7 divided by 2 is 3 with
  remainder **1**. Hugely useful: `x % 2 == 0` tells you x is even.

````python
a = 7
b = 2

print("a + b  =", a + b)
print("a - b  =", a - b)
print("a * b  =", a * b)
print("a / b  =", a / b)     # true division -> float
print("a // b =", a // b)    # floor division
print("a % b  =", a % b)     # remainder
print("a ** b =", a ** b)    # power
````

### Example 2 — Practical: splitting pizza among friends

`//` and `%` solve "how many whole X and what is left over":

````python
# 23 slices of pizza, groups of 4 students:
total_slices = 23
group_size = 4

whole_pizzas_rounds = total_slices // group_size   # full rounds of 4
leftover_slices = total_slices % group_size        # slices that remain

print("Full groups of 4:", whole_pizzas_rounds)
print("Leftover slices:", leftover_slices)
````

## 6. Comparison operators

Comparisons ask questions whose answer is `True` or `False`:

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `==` | equal to | `5 == 5` | `True` |
| `!=` | not equal to | `5 != 3` | `True` |
| `>` | greater than | `5 > 3` | `True` |
| `<` | less than | `5 < 3` | `False` |
| `>=` | greater than or equal | `5 >= 5` | `True` |
| `<=` | less than or equal | `4 <= 3` | `False` |

````python
marks = 72
pass_mark = 50

print("marks == pass_mark:", marks == pass_mark)
print("marks != pass_mark:", marks != pass_mark)
print("marks >  pass_mark:", marks > pass_mark)
print("marks <  pass_mark:", marks < pass_mark)
print("marks >= pass_mark:", marks >= pass_mark)
print("marks <= pass_mark:", marks <= pass_mark)
````

The single most common beginner bug: writing `=` (store) where `==`
(compare) is meant. `if marks = 50:` is a syntax error in Python — the

Comparisons also chain naturally: `18 <= age < 65` is legal Python and means

## 7. Logical operators

Combine boolean values into bigger questions:

| Operator | True when... | Example |
|---|---|---|
| `and` | **both** sides are True | `age >= 18 and has_id` |
| `or` | **at least one** side is True | `is_weekend or is_holiday` |
| `not` | flips the value | `not is_raining` |

````python
age = 20
has_id = True

print(age >= 18 and has_id)     # both True -> True
print(age >= 18 and not has_id) # one False -> False
print(age < 18 or has_id)       # at least one True -> True
print(not has_id)               # flip
````

**Short-circuit behaviour:** `and` stops as soon as it sees False; `or`
stops as soon as it sees True. That is why `age >= 0 and age <= 120` is safe
to evaluate quickly. Module 5 uses these operators heavily in conditions.

## 8. Assignment operators

Shortcuts for "update a variable using its own value":

| Operator | Long form | Example |
|---|---|---|
| `=` | — | `x = 5` |
| `+=` | `x = x + ...` | `x += 3` |
| `-=` | `x = x - ...` | `x -= 2` |
| `*=` | `x = x * ...` | `x *= 4` |
| `/=` | `x = x / ...` | `x /= 2` |
| `//=` | `x = x // ...` | `x //= 3` |
| `%=` | `x = x % ...` | `x %= 4` |
| `**=` | `x = x ** ...` | `x **= 2` |

````python
score = 10
score += 5     # 15
score -= 3     # 12
score *= 2     # 24
score /= 4     # 6.0  (note: /= makes a float)
score //= 2    # 3.0
score %= 2     # 1.0
score **= 3    # 1.0

print("Final score:", score)
````

## 9. Membership and identity operators — and `==` vs `is`

### Membership: `in` and `not in`

Asks whether one value is *contained* in another (strings for now; lists in
later courses):

````python
sentence = "python is fun"

print("fun" in sentence)        # True  — substring present
print("Fun" in sentence)        # False — capital F! Case matters.
print("java" not in sentence)   # True  — it is indeed absent
````

### Identity: `is` and `is not`

Asks whether two names point to the **same object in memory**.

### `==` vs `is` - the classic confusion

| | `==` | `is` |
|---|---|---|
| Question | Do the values look equal? | Are these literally the same object? |
| Use it for | Almost all comparisons | Special cases (e.g. checking `x is None`) |

````python
list_a = [1, 2, 3]
list_b = [1, 2, 3]      # same *contents*, separate object

print("list_a == list_b:", list_a == list_b)    # equal contents -> True
print("list_a is list_b:", list_a is list_b)    # different objects -> False

list_c = list_a             # both names, ONE object
print("list_a is list_c:", list_a is list_c)
````

**Rule of thumb:** use `==` to compare values (numbers, strings, ...). Use
`is` only for the special check `if value is None:`. Do **not** use `is` to
compare numbers or strings — small ints and short strings are cached by
Python for efficiency, so `is` can *appear* to work and then fail
mysteriously with bigger values.

````python
# Why `is` is wrong for values: with larger numbers the trick collapses.
# Step 1: create x — run this cell on its own.
x = 300
````

(Run the cell above, then the one below. They are *separate cells*, so
Python builds two separate `300` objects — the same thing that happens in
a fresh program. If you paste both lines into one cell, Python may
recognise the repeated constant and reuse one object, which is exactly why
`is` is unreliable for values.)

````python
# Step 2: create y and compare:
y = 300
print(x == y)    # True  — equal values
print(x is y)    # False — two separate objects that happen to hold 300
````

## 10. Expressions and statements

  `age >= 18`, `name == "Amina"`.

Every statement may *contain* expressions, but not every expression is a

## 11. Operator precedence

When an expression mixes operators, Python decides evaluation order by
**precedence** (highest first):

1. `()` — parentheses
2. `**`
3. unary `-` (negative sign)
4. `*`, `/`, `//`, `%`
5. `+`, `-`
6. comparisons (`<`, `>=`, `==`, ...)
7. `not`
8. `and`
9. `or`

Same level → evaluated left to right (`**` is right to left).

````python
# Multiplication before addition:
print(2 + 3 * 4)          # 3*4 first -> 2 + 12 -> 14

# Parentheses win:
print((2 + 3) * 4)        # 5*4 -> 20

# Power before multiplication, and ** is right-to-left:
print(2 * 3 ** 2)         # 3**2 first -> 2 * 9 -> 18
print(2 ** 3 ** 2)        # 3**2 first -> 2**9 -> 512 (NOT (2**3)**2=64)

# Comparisons after arithmetic:
print(2 + 3 > 4)          # 5 > 4 -> True
````

### Best practice: parenthesise for humans

Even when you know the rules, **add parentheses to make your intention
obvious** — your teammates (and future-you) will thank you:

````python
# Which is clearer?
average_unclear = 80 + 90 + 70 / 3        # bug! only 70 is divided
average_clear = (80 + 90 + 70) / 3

print("Wrong:", average_unclear)
print("Right:", average_clear)
````

### Handling invalid input — the basic idea

What if the user types `twenty` where a number is expected?

The beginner-friendly first line of defence is to **check before

## 4. f-strings — formatting output properly

### Concept

An **f-string** is a string with an `f` before the opening quote. Inside it,
`{...}` holds any expression — Python inserts its value into the text:

### Syntax

```python
f"text {variable} more text {expression} text"
```

````python
name = "Amina"
age = 19

print(f"Hello {name}!")
print(f"Next year {name} will be {age + 1}.")
````

f-strings beat `+`-concatenation: no `str()` calls, no spacing mistakes,
and any expression works inside `{...}`.

````python
# The old way vs f-strings:
price = 249.5
quantity = 3

old_way = "Total: " + str(price * quantity) + " PKR"
new_way = f"Total: {price * quantity} PKR"

print(old_way)
print(new_way)
````

### Formatting numbers: decimals, padding, percent

Inside the braces, a `:` starts a **format spec**. The workhorse is
`.Nf` — *N decimal places, fixed-point*:

````python
pi = 3.141592653589793

print(f"pi = {pi:.2f}")         # 2 decimal places
print(f"pi = {pi:.4f}")         # 4 decimal places
````

### More formatting in action

The same idea with shopping-style values — decimals, percent, and fixed
widths for aligned columns:

````python
pi = 3.141592653589793
total_bill = 4550.0
discount_fraction = 0.15

print(f"pi = {pi:.2f}")                       # 2 decimals
print(f"pi = {pi:.4f}")                       # 4 decimals
print(f"Bill: {total_bill:.1f} PKR")          # 1 decimal
print(f"Discount: {discount_fraction:.0%}")   # percent style -> 15%
print(f"Wide: [{total_bill:10.2f}]")          # width 10, right-aligned
print(f"Left: [{total_bill:<10.2f}]")         # width 10, left-aligned
````

Spec cheat-sheet:

| Spec | Meaning | Example → Output |
|---|---|---|
| `:.2f` | 2 fixed decimals | `f"{2.71828:.2f}"` → `2.72` |
| `:8.2f` | width 8 + 2 decimals | `f"{45.5:8.2f}"` → `   45.50` |
| `:.0%` | percentage | `f"{0.256:.0%}"` → `26%` |
| `:,` | thousands separator | `f"{2450000:,}"` → `2,450,000` |

### Common Mistakes

```python
# 1. Forgetting the f prefix — braces print literally:
name = "Ali"
print("Hello {name}")     # Hello {name}  (oops)
print(f"Hello {name}")    # Hello Ali     (correct)

# 2. Quotes inside braces:
# f"{'yes' if 5 > 3 else 'no'}" — legal, but confusing at this stage;
# compute first, then insert.
```

### Bonus — `str.format()`, the older way (you will see it in existing code)

Before f-strings existed, text was filled in with the `.format()` method: each
`{}` in the string is a **placeholder** that receives the next value.

````python
name = "Amina"
age = 19
score = 87.456

print("Dear {} your age is {}".format(name, age))                  # {} filled in order
print("name = {} and age is {}".format(name, age))
print("{0} is {1} years old; {0} scored {2:.1f}".format(name, age, score))   # numbered; format spec after ':'
print("{n} scored {s:.1f}".format(n=name, s=score))                # named placeholders

# The same lines with f-strings (preferred in new code):
print(f"{name} is {age} years old; {name} scored {score:.1f}")
````

`.format()` and f-strings produce the same text — f-strings are shorter and easier to
read, so prefer them in new programs, but recognise `.format()` when you meet it.

Why two kinds of quotes? So you can nest them:

## 2. Indexing

### Concept

Each character in a string has a position number called an **index**.
Python counts from **0**:

```text
 string:   P   y   t   h   o   n
 index:    0   1   2   3   4   5
 negative: -6  -5  -4  -3  -2  -1
```

* **Positive indexing** starts at 0 from the left.
* **Negative indexing** starts at -1 from the right — handy for "the last
  character" without knowing the length.

### Syntax

```python
string[index]
```

### Example 1 — Beginner

````python
word = "Python"

print(word[0])      # first character
print(word[1])      # second character
print(word[5])      # sixth (last) character
````

````python
print(word[-1])     # last character
print(word[-2])     # second to last
print(word[-6])     # first character, counted from the right
````

### Explanation

`word[0]` asks the string for the character at position 0 — the result is a
new, one-character string. `word[-1]` counts one from the end.

### Common Mistakes

```python
word = "Python"
# word[6]          # IndexError: string index out of range (valid: 0..5)
# word["1"]        # TypeError: indices must be integers
```

### Example 3 — Challenge: last character of any word

````python
# The last character without knowing the word in advance:
any_word = "Karachi"

last_character = any_word[len(any_word) - 1]   # len = 7, index 6
print(last_character)

print(any_word[-1])        # the idiomatic way — same result
````

## 3. Slicing

### Concept

A **slice** extracts a substring — a range of characters. This is one of
Python's most-loved features.

### Syntax

```python
string[start:stop]          # from index start up to (NOT including) stop
string[start:stop:step]     # same, taking every `step`-th character
```

**The stop index is always EXCLUDED.** `word[0:3]` gives characters at
indices 0, 1, 2 — three characters, none numbered 3.

Think of the numbers as **cut positions between the characters**:

```text
  | P | y | t | h | o | n |
  0   1   2   3   4   5   6
```

`word[1:4]` takes everything between cut 1 and cut 4 → `"yth"`.

### Example 1 — Beginner

````python
word = "Python"

print(word[0:3])     # indices 0,1,2 -> "Pyt"  (3 is excluded!)
print(word[1:4])     # indices 1,2,3 -> "yth"
print(word[3:])      # from index 3 to the end -> "hon"
print(word[:3])      # from the start up to (not incl.) 3 -> "Pyt"
print(word[:])       # the whole string -> "Python"
````

### Explanation

* Omitted `start` means "from the beginning".
* Omitted `stop` means "to the very end".
* `[:]` copies the whole string.

### Example 2 — Practical: first name, last name, extension

````python
full_name = "Amina Khan"
file_name = "report.pdf"

first_name = full_name[:5]        # "Amina"
last_name = full_name[6:]         # "Khan"
extension = file_name[-3:]        # last 3 characters: "pdf"

print("First:", first_name)
print("Last: ", last_name)
print("Type: ", extension)
````

### Negative indexes and step values

````python
word = "Python"

print(word[-4:-1])    # from -4 up to (not incl.) -1 -> "tho"
print(word[0:6:2])    # every 2nd char: indices 0,2,4 -> "Pto"
print(word[::3])      # every 3rd char: indices 0,3 -> "Ph"
print(word[::-1])     # step -1 walks backwards -> "nohtyP" (reversed!)
````

`word[::-1]` — the famous Python idiom for reversing a string. With a
negative step, a sensible reading of `word[::-1]` is "walk from the end to
the start".

### Example 3 — Challenge: middle character(s)

````python
# The middle character of an odd-length word:
word = "Karachi"      # 7 letters
middle_index = len(word) // 2
print(word[middle_index])            # 'a' — index 3

# For even length, grab the two middle characters:
even_word = "Lahore"         # 6 letters
centre = len(even_word) // 2            # 6 // 2 = 3
print(even_word[centre - 1:centre + 1]) # indices 2 and 3 -> "ho"
````

### Slicing never raises IndexError

Unlike indexing, a slice that runs past the end is silently clipped:
`"Python"[3:100]` gives `"hon"`. Handy — but it can hide typos.

## 4. String operations

### Concatenation `+`

Joins strings into a new string. Both sides must be strings:

````python
first_part = "Data"
second_part = "Science"

joined = first_part + " " + second_part
print(joined)
````

````python
# "+" does NOT convert: gluing a number needs str() or an f-string
age = 19
print("Age: " + str(age))
print(f"Age: {age}")
````

### Repetition `*`

A string times an integer repeats it:

````python
divider = "-" * 30
print(divider)
print("Report card")
print(divider)
````

### Membership `in` / `not in`

Checks whether a substring occurs anywhere in the string (True/False):

````python
sentence = "Python is a popular language"

print("popular" in sentence)
print("Popular" in sentence)      # case matters!
print("java" not in sentence)
````

### Comparison `==`, `<`, `>` ...

Strings compare by **dictionary/alphabetical order** (technically by
Unicode code points — so all uppercase letters sort before lowercase):

````python
print("apple" == "apple")
print("apple" < "banana")
print("Zebra" < "apple")     # True: 'Z' (90) < 'a' (97)!
print("abc" == "ABC")        # False: case-sensitive
````

Practical consequence: before comparing user input, normalise case
(`.lower()`) — see section 7.

## 5. String immutability

### Concept

Strings are **immutable**: they cannot be changed after creation. Every
"modification" actually builds a **new** string.

### Example — what fails and what works

````python
word = "cat"
# The following line WOULD fail — try uncommenting it:
# word[0] = "b"        # TypeError: 'str' object does not support item assignment

new_word = "b" + word[1:]     # build a NEW string instead
print(new_word)
````

````python
# Even method calls return new strings; the original is untouched:
shout = "hello"
upper_shout = shout.upper()

print(shout)        # still "hello"
print(upper_shout)  # the new string "HELLO"
````

**Why it matters:** always *capture the result*: `name = name.strip()` —
not just `name.strip()` (which does nothing visible!).

## 1. Boolean logic — the foundation

Before writing any `if`, be fluent with the **booleans** it depends on.

* A boolean value is either `True` or `False` (the only two, capitalised!).
* A **boolean expression** is any expression whose value is `True` or
  `False` — usually built with **comparison operators**.

| Comparison | Meaning |
|---|---|
| `==` | equal to |
| `!=` | not equal |
| `<`, `<=` | less (or equal) |
| `>`, `>=` | greater (or equal) |

````python
print(10 > 3)          # True
print(10 == 3)         # False
print(10 != 3)         # True
marks = 65
print(marks >= 50)     # boolean expression — evaluates to True
````

### Logical operators: `and`, `or`, `not`

They combine boolean expressions:

| `a` | `b` | `a and b` | `a or b` | `not a` |
|---|---|---|---|---|
| True | True | **True** | True | False |
| True | False | False | **True** | False |
| False | True | False | **True** | True |
| False | False | False | False | True |

* `and` — True only when **both** sides are True.
* `or` — True when **at least one** side is True.
* `not` — flips the value.

````python
age = 20
has_id = True

print(age >= 18 and has_id)     # both True          -> True
print(age < 18 or has_id)       # one True is enough -> True
print(not has_id)               # flip               -> False
````

## 2. The `if` statement

### Concept

Programs need to **decide**: do this *only if* something is true. The `if`
statement runs its indented block exactly when its condition is `True`;
otherwise the block is skipped and the program continues.

### Syntax

```python
if condition:
    statement          # runs when condition is True
    statement          # (any number of indented lines)
statement_after       # ALWAYS runs — not indented
```

Note the **colon `:`** ending the `if` line and the **4-space indentation**
of the block.

### Example 1 — Beginner

````python
temperature = 35

if temperature > 30:
    print("It is hot today.")
    print("Drink plenty of water.")

print("(This line is outside the if — it always runs.)")
````

### Explanation

Line by line:

* `temperature = 35` — store 35.
* `if temperature > 30:` — evaluate `35 > 30` → `True`, so the indented
  block below runs.
* The last `print` is **not indented**, so it is outside the `if` and runs
  no matter what.

Change `temperature` to `25` and run again: the two inside lines vanish,
the outside line stays.

````python
temperature = 25          # same code, new value — now the block is skipped

if temperature > 30:
    print("It is hot today.")

print("(outside — always runs)")
````

### Example 2 — Practical: a low-stock warning

````python
stock = 4

if stock < 5:
    print("WARNING: only", stock, "items left — reorder soon!")
````

## 3. `if ... else` — two paths

### Concept

Often exactly one of two things must happen. `else` catches the "not `if`"
case. The `else` has **no condition** of its own.

### Syntax

```python
if condition:
    ...            # runs when condition is True
else:
    ...            # runs when condition is False
```

### Example 1 — Beginner: even or odd

````python
number = 7

if number % 2 == 0:
    print(number, "is even")
else:
    print(number, "is odd")
````

### Explanation

* `number % 2` is the remainder after dividing by 2 — `0` for even numbers,
  `1` for odd.
* The condition `number % 2 == 0` compares that remainder with 0.
* For 7: remainder 1 → condition `False` → the `else` block runs.

### Example 2 — Practical: pass/fail

````python
marks = 42

if marks >= 50:
    print("Result: PASS")
else:
    print("Result: FAIL")
````

## 4. `if ... elif ... else` — many paths

### Concept

`elif` (short for *else if*) adds more conditions. Python checks each
condition **top to bottom** and runs the block of the **first** one that is
True — everything after is skipped. `else` (optional) catches all remaining
cases.

### Syntax

```python
if condition_one:
    ...
elif condition_two:
    ...
elif condition_three:
    ...
else:
    ...
```

**Order matters!** Conditions are tested top to bottom, so put the most
specific/strictest first (in grading: the highest grade).

### Example 1 — Beginner: grade calculator

````python
marks = 78

if marks >= 80:
    grade = "A"
elif marks >= 70:
    grade = "B"
elif marks >= 60:
    grade = "C"
elif marks >= 50:
    grade = "D"
else:
    grade = "F"

print("Marks:", marks, "-> Grade:", grade)
````

### Explanation

For 78: `78 >= 80`? No. `78 >= 70`? **Yes** → `grade = "B"`, and the other
branches are not even checked. Why must `A` be tested first? If `>= 50`
were first, every mark ≥ 50 would get D — the first matching branch wins.

### Example 2 — Practical: electricity bill with slabs

````python
# First 100 units: 5.0/unit; next 100: 8.0/unit; above 200: 12.0/unit.
units = 250

if units <= 100:
    bill = units * 5.0
elif units <= 200:
    bill = 100 * 5.0 + (units - 100) * 8.0
else:
    bill = 100 * 5.0 + 100 * 8.0 + (units - 200) * 12.0

print(f"Units: {units}, Bill: {bill:.0f} PKR")
````

### Explanation

Each branch builds on the previous slabs: the first 100 units always cost
5 each, the next 100 cost 8 each, only units beyond 200 cost 12. For 250
units: 500 + 800 + 600 = 2100.

## 5. Indentation — the rules, carefully

Python uses indentation to define blocks. Four rules to memorise:

1. Use **4 spaces** per level (PEP 8). Configure your editor to insert
   spaces when Tab is pressed.
2. Every line in the same block must have the **same** indentation.
3. **Deeper** indent = inside the block; returning to a shallower indent
   ends the block.
4. Never mix tabs and spaces (Python 3 raises `TabError`).

```python
if marks >= 50:
    print("Pass")           # inside: 4 spaces
    print("Well done")      # inside: same 4 spaces
print("Done")               # outside: no indent — runs always
```

**Errors you will meet:**

```python
if marks >= 50:
print("Pass")        # IndentationError: expected an indented block

    print("Pass")    # IndentationError: unexpected indent (no block above)
```

Nested blocks simply indent deeper (8 spaces, then 12, ...):

````python
age = 19
has_national_id = True

if age >= 18:
    print("Adult check:")
    if has_national_id:                       # nested if: indented 4 more
        print("  Eligible to vote.")          # 8 spaces
    else:
        print("  Get your ID card first.")
````

## 1. Why loops?

### Concept

Suppose you must print "Happy Birthday!" ten times. Copy-pasting the same
line ten times works but is ugly, error-prone, and impossible for values
you only learn at run time ("print it 500 times", "print it once per
student"). **Loops** repeat a block of code automatically. One execution of
the block is called an **iteration**; running a loop is called
**iterating**; the repeated block is the **loop body**.

Manual repetition vs a loop:

````python
# Manual repetition — copy, paste, pray:
print("Happy Birthday!")
print("Happy Birthday!")
print("Happy Birthday!")

print("--- with a loop ---")

# A loop does the same in three lines, for ANY count:
for day_number in range(3):
    print("Happy Birthday!")
````

Change `range(3)` to `range(500)` and the loop happily does 500 iterations
— the manual version would need 500 pasted lines. That is why loops exist.

Python has two loop statements:

* **`for`** — repeat **once per item** of a sequence (a known count, a
  string's characters, ...). Use when you know *what* to iterate over.
* **`while`** — repeat **as long as** a condition stays True. Use when you
  know *when to stop*, but not how many rounds it will take.

## 2. The `for` loop and `range()`

### Syntax

```python
for variable in sequence:
    body                # indented 4 spaces, runs once per item
```

`range()` produces number sequences — three forms:

| Form | Meaning | Example |
|---|---|---|
| `range(stop)` | 0 up to stop−1 | `range(5)` → 0,1,2,3,4 |
| `range(start, stop)` | start up to stop−1 | `range(2, 6)` → 2,3,4,5 |
| `range(start, stop, step)` | every `step`-th | `range(0, 10, 2)` → 0,2,4,6,8 |

Like slices, the **stop value is excluded**.

### Example 1 — Beginner

````python
# Prints 0 1 2 3 4 — five iterations, stop value 5 is NOT printed.
for number in range(5):
    print(number)
````

### Explanation

Line by line:

* `for number in range(5):` — create the sequence 0,1,2,3,4. On each
  iteration, the next value is stored in `number` (the **loop variable**).
* `print(number)` — the body; runs once per value.
* The loop ends when the sequence is exhausted; execution continues below
  the loop.

### The three range forms in action

````python
print("range(5):")
for number in range(5):
    print(number, end=" ")
print()                              # move to a new line

print("range(2, 6):")
for number in range(2, 6):           # 2,3,4,5
    print(number, end=" ")
print()

print("range(0, 10, 2):")
for number in range(0, 10, 2):       # 0,2,4,6,8
    print(number, end=" ")
print()

print("range(5, 0, -1):")            # counting DOWN
for number in range(5, 0, -1):       # 5,4,3,2,1
    print(number, end=" ")
print()
````

**Note:** `range(5, 0, -1)` counts down but never reaches 0 — the stop is
excluded. Use `range(5, -1, -1)` to include it.

## 4. The `while` loop

### Syntax

```python
while condition:
    body            # repeats while condition stays True
```

**The three ingredients of a well-behaved `while` loop:**

1. **Initialise** the control variable *before* the loop.
2. **Test** it in the condition.
3. **Update** it inside the body — or the loop never ends!

### Example 1 — Beginner: counting up

````python
count = 1                      # 1. initialise

while count <= 5:              # 2. test
    print("Count is", count)
    count += 1                 # 3. update — without this: infinite loop!
````

### Explanation

* Round 1: `count` is 1 → `1 <= 5` True → print, `count` becomes 2.
* ... continues until `count` is 6 → `6 <= 5` False → loop ends.
* The body ran exactly 5 times.

### Example 2 — Practical: saving until a goal is reached

````python
balance = 0
monthly_saving = 2500
target = 10000
months = 0

while balance < target:
    balance += monthly_saving
    months += 1

print(f"Target reached in {months} months with {balance} PKR")
````

A `while` is perfect here: we cannot know in advance how many months it
will take — the loop decides by itself.

### Infinite loops — and how to avoid them

````python
# DO NOT RUN the next three lines as-is: they never stop.
# while True:
#     print("forever")      # condition is always True, nothing changes it

# The fix: ensure SOMETHING in the body can make the condition False.
countdown = 3
while countdown > 0:
    print(countdown)
    countdown -= 1         # the escape hatch

print("Done")
````

If a notebook cell ever runs forever: **Kernel → Interrupt** (square stop
button) stops the runaway loop safely.

# value = 0.0

### 6.1 Counting — how many times does something happen?

````python
count = 0                       # the counter starts at 0

for number in range(1, 51):
    if number % 7 == 0:         # multiple of 7?
        count += 1              # one more found

print("Multiples of 7 up to 50:", count)
````

### 6.3 Product — multiplying things together

Start at **1** (multiplying by 0 would erase everything):

````python
product = 1

for factor in (2, 3, 4):
    product *= factor

print("2 * 3 * 4 =", product)
````

### 6.4 Minimum and maximum

Carry the best-so-far; replace it whenever a better value appears:

````python
temperatures = (31, 28, 35, 33, 29)

highest = temperatures[0]        # start with the first value...
lowest = temperatures[0]

for temperature in temperatures:
    if temperature > highest:    # ...then challenge it each round
        highest = temperature
    if temperature < lowest:
        lowest = temperature

print("Highest:", highest)
print("Lowest:", lowest)
````

**Why start with the first element?** Starting `highest` at 0 fails when
all values are negative; starting it at 999999 fails for hot days. The
first element is always a safe starting champion.

### 6.5 Average — sum and count together

````python
marks = (78, 92, 65, 88)

marks_total = 0
marks_count = 0

for mark in marks:
    marks_total += mark
    marks_count += 1

average = marks_total / marks_count
print(f"Average of {marks_count} marks: {average:.2f}")
````

### 6.6 Frequency — counting categories

````python
sentence = "abracadabra"

count_a = 0
count_b = 0
count_other = 0

for character in sentence:
    if character == "a":
        count_a += 1
    elif character == "b":
        count_b += 1
    else:
        count_other += 1

print("a:", count_a, "| b:", count_b, "| other:", count_other)
````

## 7. Nested loops

A loop **inside** a loop. For every single step of the outer loop, the

### 8.2 Reverse a number

Module 4 reversed text with slicing; numbers need arithmetic — the digit

````python
number = 4793
reversed_number = 0
remaining = number

while remaining > 0:
    last_digit = remaining % 10            # take the last digit...
    reversed_number = reversed_number * 10 + last_digit   # ...append it
    remaining //= 10                       # chop it off

print(number, "reversed is", reversed_number)
````

**Trace for 4793:** digit 3 → rev=3, remaining=479 → digit 9 → rev=39,

### 8.4 Prime number check

A **prime** has exactly two divisors: 1 and itself. Strategy: try every
divisor from 2 up to number−1; if any divides evenly, it is *not* prime:

````python
number = 29
is_prime = True                    # innocent until proven guilty

for divisor in range(2, number):
    if number % divisor == 0:      # found another divisor!
        is_prime = False
        break                      # no need to keep looking

if number < 2:
    is_prime = False               # 0 and 1 are not prime

print(number, "is prime:", is_prime)
````

````python
# All primes below 30 — nested logic + accumulator + conditions:
print("Primes below 30:")
for candidate in range(2, 30):
    is_prime = True
    for divisor in range(2, candidate):
        if candidate % divisor == 0:
            is_prime = False
            break
    if is_prime:
        print(candidate, end=" ")
print()
````

### 8.5 Fibonacci sequence

Each Fibonacci number is the sum of the two before it: 0, 1, 1, 2, 3, 5,
8, ... The trick is updating **two** variables per round:

````python
terms = 8
previous = 0
current = 1

for term_number in range(terms):
    print(previous, end=" ")
    next_value = previous + current   # compute the next BEFORE overwriting
    previous = current                # shift the window forward
    current = next_value
print()
````

**Order matters:** if you wrote `previous = current` *before* computing
`next_value`, you would lose the old `previous`. Compute first, shift
second.

### 8.6 Find maximum and minimum with a loop

Section 6.4 showed the pattern on a tuple; here the values arrive one by
one inside the loop (like readings from a sensor):

````python
# Simulated readings — the loop must handle each as it comes:
readings = (21, 19, 24, 30, 17, 24)

highest_so_far = None      # nothing seen yet
lowest_so_far = None

for reading in readings:
    if highest_so_far is None or reading > highest_so_far:
        highest_so_far = reading
    if lowest_so_far is None or reading < lowest_so_far:
        lowest_so_far = reading

print("Max:", highest_so_far, "| Min:", lowest_so_far)
````

(This is also the first honest use of `is None` from Module 2!)

## 3. Design — the algorithm in plain words

## 4. Pseudocode

Pseudocode is the algorithm written in structured half-English/half-code —
precise about the *logic*, silent about the *syntax*:

```text
START
  DISPLAY banner
  running <- "YES"
  WHILE running = "YES":
      name <- INPUT "Student name"
      score <- 0
      FOR each (question, options, correct) IN quiz_data:
          DISPLAY question
          DISPLAY options
          answer <- INPUT "Your answer"
          answer <- UPPERCASE( TRIM(answer) )
          IF answer = correct THEN
              score <- score + 1
              DISPLAY "Correct!"
          ELSE
              DISPLAY "Wrong. Correct answer:", correct
          ENDIF
      ENDFOR
      percentage <- score / 5 * 100
      IF percentage >= 80 THEN grade <- "A"
      ELIF percentage >= 60 THEN grade <- "B"
      ELIF percentage >= 40 THEN grade <- "C"
      ELSE grade <- "F"
      ENDIF
      DISPLAY report(name, score, percentage, grade)
      running <- UPPERCASE( TRIM( INPUT "Another student? yes/no" ) )
  ENDWHILE
  DISPLAY "Goodbye"
END
```

Notice how the pseudocode already names the patterns you know: a `WHILE`
sentinel loop around everything, a `FOR` loop over the questions, a counter
`score`, and an `IF/ELIF` chain for the grade.

## 5. The quiz data — one gentle new idea

Five questions, each with its text, its options, and its correct letter.
Python's **tuple** (introduced as "a sequence" in Modules 2 and 6) lets us
group each question's pieces:

````python
question_one = ("Which function displays text on the screen?",
                "A. print()   B. show()   C. display()",
                "A")
question_two = ("What does input() always return?",
                "A. an integer   B. a string   C. a float",
                "B")
question_three = ("Which operator gives the remainder?",
                  "A. /   B. //   C. %",
                  "C")
question_four = ("Which keyword starts a conditional statement?",
                 "A. if   B. for   C. while",
                 "A")
question_five = ("How many times does range(3) loop?",
                 "A. 2   B. 3   C. 4",
                 "B")

# A tuple of five tuples: the whole quiz in one variable.
quiz_data = (question_one, question_two, question_three,
             question_four, question_five)

print("Questions loaded:", len(quiz_data))
````

Each item is used exactly as in Module 6's loops: `for question in
quiz_data:` visits every question in order, and `question[0]`,
`question[1]`, `question[2]` are its text, options and correct letter.

## 7. The complete program

Everything assembled, exactly as designed. This cell contains **real
`input()` calls** — when you run it in Jupyter, type into the input boxes:
name first, then five answers (A/B/C), then yes/no. To replay the exact
session shown in the sample run, type: `Amina`, then `a`, `b`, `c`, `a`,
`b`, then `no`.

````python
# ============================================================
#  MINI PROJECT: Python Basics Self-Test with Grade Report
#  Uses only: variables, types, I/O, strings, conditions, loops
# ============================================================
# This cell is fully interactive. A sample session to try:
# Amina, a, b, c, a, b, no

print("=" * 46)
print("PYTHON BASICS SELF-TEST   (5 questions)".center(46))
print("=" * 46)

running = "YES"

while running == "YES":

    # ----- R1: student name -----
    student_name = input("Student name: ")
    print()
    print(f"Welcome, {student_name}! Answer with A, B or C.")
    print()

    # ----- R5: the score counter starts at zero -----
    score = 0

    # ----- R2: ask every question in order -----
    for question, options, correct in quiz_data:
        print("-" * 46)
        print(question)
        print(options)
        raw_answer = input("Your answer: ")

        # ----- R3: clean the answer -----
        clean_answer = raw_answer.strip().upper()

        # ----- R4: immediate feedback -----
        if clean_answer == correct:
            print("Correct!")
            score += 1
        else:
            print(f"Wrong — the correct answer was {correct}.")

    # ----- R6: percentage and grade -----
    total_questions = len(quiz_data)
    percentage = score / total_questions * 100

    if percentage >= 80:
        grade = "A"
    elif percentage >= 60:
        grade = "B"
    elif percentage >= 40:
        grade = "C"
    else:
        grade = "F"

    passed = percentage >= 40

    # ----- R7: the grade report -----
    print()
    print("=" * 46)
    print("            GRADE REPORT")
    print("=" * 46)
    print(f"  Student    : {student_name}")
    print(f"  Score      : {score} out of {total_questions}")
    print(f"  Percentage : {percentage:.0f}%")
    print(f"  Grade      : {grade}")
    print(f"  Result     : {'PASS' if passed else 'FAIL'}")

    if percentage == 100:
        feedback = "Perfect score — outstanding!"
    elif percentage >= 80:
        feedback = f"Excellent work, {student_name}!"
    elif percentage >= 60:
        feedback = f"Good effort, {student_name} — review the misses."
    elif percentage >= 40:
        feedback = f"You passed, {student_name}, but revisit Modules 2-4."
    else:
        feedback = f"Keep practising, {student_name} — try the quiz again."
    print(f"  Feedback   : {feedback}")
    print("=" * 46)

    # ----- R8: run again? -----
    running = input("Run for another student? (yes/no): ").strip().upper()

print()
print("Thank you for using the self-test. Goodbye!")
````

### How the code maps to the design

| Design step | Code |
|---|---|
| 2. `running` set to "YES" | `running = "YES"` before the `while` |
| 3a. read name | `student_name = input(...)` |

forever — the classic `while` mistake from Module 6.

## 8. Testing the program

Good tests try *different kinds* of answers, not just the happy path:

| Test | Input answer | Cleaned | Expected effect |
|---|---|---|---|
| lowercase | `b` | `B` | accepted as B |
| padded | `"  a "` | `A` | accepted as A |
| wrong case+spaces | `" c  "` | `C` | accepted as C |
| invalid letter | `x` | `X` | counted wrong, program continues |
| empty | `` (just Enter) | `` | counted wrong, program continues |
| stop signal | `no` at the prompt | `NO` | program ends politely |

Notice the design choice: invalid answers are simply *wrong*, the quiz
never crashes. Robust programs degrade gracefully.

## Types and conversion

| Type | Example | Convert with |
|---|---|---|
| `int` | `42` | `int("42")`, `int(3.9)` → `3` (cuts decimals) |
| `float` | `3.14` | `float("3.14")`, `float(7)` |
| `str` | `"text"` | `str(42)` |
| `bool` | `True`, `False` | `bool(0)` → `False`, `bool("hi")` → `True` |
| `NoneType` | `None` | “no value yet” — test with `x is None` |

`type(x)` tells you the type of any value.

## Operators

| Kind | Operators |
|---|---|
| Arithmetic | `+` `-` `*` `/` (always float) `//` (floor) `%` (remainder) `**` (power) |
| Comparison | `==` `!=` `>` `<` `>=` `<=` (chaining allowed: `18 <= age < 65`) |
| Logical | `and` `or` `not` |
| Assignment | `=` `+=` `-=` `*=` `/=` `//=` `%=` `**=` |
| Membership / identity | `in` `not in` / `is` `is not` (use `==` for values, `is` only for `None`) |

Precedence (high → low): `()` → `**` → unary `-` → `* / // %` → `+ -` → comparisons → `not` → `and` → `or`.

## Input and output

| Task | Code |
|---|---|
| Print values | `print(a, b)` — options `sep=" "`, `end="\n"` |
| Read text | `text = input("Prompt: ")` — **always a string** |
| Read a number | `n = int(input("..."))` or `float(input("..."))` |
| f-string | `f"{name} scored {marks:.1f}"` |
| Number formats | `{x:.2f}` 2 decimals · `{x:,}` thousands · `{x:.0%}` percent · `{x:>10}` right-align · `{x:<10}` left-align |
| Escapes | `\n` newline · `\t` tab · `\\` backslash · `\"` quote |

## Strings

| Task | Code |
|---|---|
| Length | `len(s)` |
| Index / last char | `s[0]`, `s[-1]` |
| Slice | `s[a:b]` (b excluded), `s[a:b:step]`, `s[::-1]` reverses |
| Join / repeat | `s1 + s2`, `s * 3` |
| Case | `lower()` `upper()` `capitalize()` `title()` |
| Clean | `strip()` `lstrip()` `rstrip()` |
| Change | `replace(old, new)` |
| Split / join | `"a,b".split(",")` → list · `"-".join(list)` |
| Search | `find()` (−1 if absent) · `index()` (error if absent) · `count()` · `startswith()` · `endswith()` |

Strings are **immutable** — methods return a *new* string, so always capture the result: `s = s.strip()`.

## Loops

```python
for item in range(start, stop, step):   # stop is excluded
    ...

while condition:        # initialise → test → UPDATE
    ...
```

| Keyword | Effect |
|---|---|
| `break` | leave the loop now |
| `continue` | skip to the next iteration |
| `pass` | do nothing (placeholder) |

Patterns: **counter** (`count += 1`) · **sum** (`total += x`, start 0) · **product** (`p *= x`, start 1) ·
**max/min** (start with the first value) · **sentinel** (a special value that stops the loop).
